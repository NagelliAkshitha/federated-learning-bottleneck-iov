```python
import cv2
import numpy as np
# For displaying images in Colab
# If you are not in Colab, you might need to use cv2.imshow instead
from google.colab.patches import cv2_imshow

# --- CONFIGURATION ---
# Change this to your uploaded image path
image_path = "/content/img 8.jpg"
# Number of clusters for K-Means
K_CLUSTERS = 4
# --- END CONFIGURATION ---

# Step 1 : Read the input image
img = cv2.imread(image_path)

if img is None:
    raise ValueError(f"Image not found. Check the file path: {image_path}")

print("Original Image")
cv2_imshow(img)


# Step 2 : Detect faces using Haar Cascade
# Load the pre-trained Haar Cascade classifier for frontal faces
try:
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
    if face_cascade.empty():
        raise IOError("Could not load Haar cascade classifier. Check your OpenCV installation.")
except Exception as e:
    print(f"Error loading Haar cascade: {e}")
    print("Please ensure you have the 'opencv-python' package installed correctly.")
    # As a fallback, you might need to download the file manually if cv2.data.haarcascades isn't populated
    # and provide the full path to 'haarcascade_frontalface_default.xml'
    exit()



faces = face_cascade.detectMultiScale(
    img,
    scaleFactor=1.1,
    minNeighbors=8,  # Increased from 5 to 8
    minSize=(30, 30)
)

if len(faces) == 0:
    print("No faces detected in the image.")
else:
    print(f"\nDetected {len(faces)} face(s). Processing each...\n")

    # --- MODIFICATION: Loop over all detected faces ---
    for (x, y, w, h) in faces:

        # Draw a rectangle on the original image to show which face is being processed
        img_with_rect = img.copy()
        cv2.rectangle(img_with_rect, (x, y), (x+w, y+h), (0, 255, 0), 2)
        print("--- Processing Face ---")
        cv2_imshow(img_with_rect)

        # Crop the first detected face
        face = img[y:y+h, x:x+w]

        # Ensure face has valid dimensions
        if face.shape[0] == 0 or face.shape[1] == 0:
            print("Skipping a detected region with zero area.")
            continue

        # Step 3 : Convert image to 2D array of pixels
        # Reshape to a (num_pixels, 3) array (3 for B,G,R channels)
        pixel_values = face.reshape((-1, 3))
        pixel_values = np.float32(pixel_values)

        # Step 4 : Define criteria and number of clusters
        # Stop iterating when epsilon (1.0) is reached or max_iter (10) is reached
        criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 10, 1.0)
        k = K_CLUSTERS

        # Step 5 : Apply K-Means clustering
        # labels: 1D array, each pixel gets a cluster ID (0 to k-1)
        # centers: (k, 3) array, holding the BGR color for each cluster
        _, labels, centers = cv2.kmeans(pixel_values, k, None, criteria, 10, cv2.KMEANS_RANDOM_CENTERS)

        # Step 6 : Convert centers to uint8 (valid color values)
        centers = np.uint8(centers)

        # Step 7 : Replace pixels with their corresponding cluster centers
        # Map each pixel's label to its center color
        segmented_pixels = centers[labels.flatten()]
        # Reshape the 1D pixel array back to the original face's shape
        segmented_image = segmented_pixels.reshape(face.shape)

        # Step 8 : Display segmented face image
        print(f"K-Means segmentation (k={k})")
        cv2_imshow(segmented_image)

        # Step 9 : Display each cluster (segment) separately
        print("--- Individual Segments ---")
        for i in range(k):
            # Create a mask for the current cluster 'i'
            # This is a 1D boolean array
            mask_1d = (labels.flatten() == i)

            # Create a new black image (3 channels) with the same shape as the face
            cluster_img = np.zeros(face.shape, dtype=np.uint8)

            # Reshape the 1D mask to match the 2D shape of the image (y, x)
            mask_2d = mask_1d.reshape(face.shape[0], face.shape[1])

            # Use the 2D mask to set the color
            # Where mask_2d is True, set the pixel color to the center color for this cluster
            cluster_img[mask_2d] = centers[i]

            cv2_imshow(cluster_img)
            print(f"Segment {i+1}")

        print("----------------------------\n")

print("All processing finished.")
```
